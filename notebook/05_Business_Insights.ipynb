```python
# ==========================================
# Cell 1 - Import Libraries
# ==========================================

import pandas as pd

pd.set_option("display.max_columns", None)

print("Libraries Imported Successfully")
```

```python
# ==========================================
# Cell 2 - Load Dataset
# ==========================================

df = pd.read_csv("../data/processed/Nassau_Candy_Cleaned.csv")

print(df.shape)
```

```python
# ==========================================
# Cell 3 - Executive Summary
# ==========================================

summary = pd.DataFrame({
    "Metric": [
        "Total Orders",
        "Total Sales",
        "Total Gross Profit",
        "Average Lead Time"
    ],
    "Value": [
        len(df),
        f"${df['Sales'].sum():,.2f}",
        f"${df['Gross Profit'].sum():,.2f}",
        f"{df['Lead Time (Days)'].mean():.2f} Days"
    ]
})

summary
```

# Executive Summary

This project evaluates the shipping performance of Nassau Candy Distributor by
analyzing factory-to-customer delivery routes.

The analysis identifies efficient routes, shipping bottlenecks,
regional performance differences, and ship mode effectiveness.

The generated insights support operational improvements,
better logistics planning, and enhanced customer satisfaction.

```python
# ==========================================
# Cell 4 - Top Factory
# ==========================================

top_factory = (
    df.groupby("Factory")
      .size()
      .sort_values(ascending=False)
)

top_factory
```

## Business Insight

The factory with the highest shipment volume plays a major role
in the distribution network.

Monitoring its operational performance is critical because
small improvements can significantly impact nationwide deliveries.

```python
# ==========================================
# Cell 5 - Best Ship Mode
# ==========================================

ship_mode = (
    df.groupby("Ship Mode")["Lead Time (Days)"]
      .mean()
      .sort_values()
)

ship_mode
```

## Business Insight

Premium shipping modes consistently demonstrate lower delivery times.

Customers requiring urgent deliveries should be encouraged
to select faster shipping options.

```python
# ==========================================
# Cell 6 - Highest Shipment States
# ==========================================

state_orders = (
    df.groupby("State/Province")
      .size()
      .sort_values(ascending=False)
)

state_orders.head(10)
```

```python
# ==========================================
# Cell 7 - Route Performance
# ==========================================

MIN_SHIPMENTS = 10  # filter out low-volume routes so single orders don't dominate

route_summary = (
    df.groupby("Route State")
      .agg(
          Shipments=("Order ID","count"),
          Avg_Lead_Time=("Lead Time (Days)","mean")
      )
)

route_summary = route_summary[route_summary["Shipments"] >= MIN_SHIPMENTS]
route_summary = route_summary.sort_values("Avg_Lead_Time")

route_summary.head(10)
```

## Business Insight

Highly efficient routes provide a benchmark for
improving slower routes.

Standardizing successful logistics practices
can improve network-wide performance.

# Recommendations

Based on the analysis, the following recommendations are proposed:

1. Optimize high-volume routes to reduce transportation delays.

2. Expand the use of faster shipping modes for priority customers.

3. Monitor factories with the largest shipment volumes to prevent operational bottlenecks.

4. Allocate additional transportation resources to high-demand states.

5. Continuously track shipping KPIs using the Streamlit dashboard.

- Future studies should incorporate actual shipment timestamps and transportation distances.
