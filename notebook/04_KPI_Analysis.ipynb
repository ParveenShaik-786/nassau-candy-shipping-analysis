```python
# ==========================================
# Cell 1 - Import Libraries
# ==========================================

import pandas as pd
import matplotlib.pyplot as plt
import plotly.express as px

pd.set_option("display.max_columns", None)

print("Libraries imported successfully.")
```

```python
# ==========================================
# Cell 2 - Load Dataset
# ==========================================

df = pd.read_csv("../data/processed/Nassau_Candy_Cleaned.csv")

print(df.shape)
```

```python
df["Order Date"] = pd.to_datetime(df["Order Date"])
df["Corrected Ship Date"] = pd.to_datetime(df["Corrected Ship Date"])
```

```python
# ==========================================
# KPI 1
# ==========================================

total_shipments = len(df)

print(f"Total Shipments : {total_shipments}")
```

```python
# ==========================================
# KPI 2
# ==========================================

avg_lead = df["Lead Time (Days)"].mean()

print(f"Average Lead Time : {avg_lead:.2f} Days")
```

```python
# ==========================================
# KPI 3
# ==========================================

delay_percentage = (
    (df["Lead Time (Days)"] > 5).mean()
) * 100

print(f"Delay Frequency : {delay_percentage:.2f}%")
```
